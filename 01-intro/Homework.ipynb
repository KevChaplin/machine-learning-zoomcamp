```python
asian_cars = df[
    df['origin'] == 'Asia'
]
X = asian_cars[['vehicle_weight', 'model_year']].head(7)
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inverse = np.linalg.pinv(XTX)
XTX_inverse
```

```python
X_transpose = X.T
XTXiXT = XTX_inverse.dot(X_transpose)
XTXiXT
```
